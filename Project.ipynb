```python
import numpy as np
import matplotlib.pyplot as plt

print("Bridge Load Distribution Calculator")
```

```python
# Bridge inputs
L = float(input("Enter bridge length L (m): "))
P = float(input("Enter load P (kN): "))
x = float(input("Enter load position from left support x (m): "))

print("\nInputs:")
print("Bridge length =", L, "m")
print("Load =", P, "kN")
print("Load position =", x, "m")
```

```python
# Create AX = B

A = np.array([
    [1, 1],
    [0, L]
], dtype=float)

B = np.array([
    [P],
    [P * x]
], dtype=float)

# Calculate inverse of A
A_inverse = np.linalg.inv(A)

# X = A^-1 B
X = A_inverse @ B

RA = X[0, 0]
RB = X[1, 0]

print("\nMatrix A:")
print(A)

print("\nMatrix B:")
print(B)

print("\nInverse of A:")
print(A_inverse)

print("\nSupport reactions:")
print("RA =", round(RA, 2), "kN")
print("RB =", round(RB, 2), "kN")
```

```python
# Visualize the bridge

plt.figure(figsize=(10, 4))

# Bridge
plt.plot([0, L], [0, 0], linewidth=6)

# Supports
plt.scatter([0, L], [0, 0], s=150)

# Load arrow
plt.arrow(
    x, 1.5,
    0, -1.2,
    head_width=0.25,
    head_length=0.25,
    length_includes_head=True
)

# Reaction arrows
plt.arrow(
    0, 0,
    0, 1.2,
    head_width=0.25,
    head_length=0.25,
    length_includes_head=True
)

plt.arrow(
    L, 0,
    0, 1.2,
    head_width=0.25,
    head_length=0.25,
    length_includes_head=True
)

# Labels
plt.text(0, -0.35, f"A\nRA = {RA:.2f} kN", ha="center")
plt.text(L, -0.35, f"B\nRB = {RB:.2f} kN", ha="center")
plt.text(x, 1.6, f"Load = {P} kN", ha="center")

plt.xlim(-1, L + 1)
plt.ylim(-1, 2.5)
plt.xlabel("Distance (m)")
plt.ylabel("Relative position")
plt.title("Bridge Load Distribution")
plt.grid(True)

plt.show()
```
